### B - Data Structures: Lists, Tuples, Sets and Dictionaries

```python
top_movies = ["Green Book", "Good Will Hunting", "LOTR: Fellowship Of The Ring", "LOTR: The Two Towers", "LOTR: The Return Of The King"]
print(top_movies)

# add a new movie at the end of the list 
top_movies.append("American Phsyco")
print(top_movies)

#remove the second movie in the list 
#del top_movies[1] deletes the value  
top_movies.pop(1) # pops it out, can be stored in a new variable and be used again
print(top_movies)
```

```python
a_list = [1, 2, 3, 4, 8, 5, 0, 3, 5]
b_list = [9, 7, 6, 7, 5, 3, 5, 4, 2]

a_set = set(a_list)
b_set = set(b_list)


print(a_set)
print(b_set)

# intersection 
intersection = a_set and b_set
print(intersection)

```
