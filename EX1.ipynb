3. Create two lists of numbers with some overlap. <br>Convert them to sets, find the intersection
(common elements), union (all unique elements), and difference. <br>Also create a list of tuples
pairing each number with its square. Use comprehension. <br>Print each result

```python
a_list = [1, 2, 3, 4, 8, 5, 0, 3, 5]
b_list = [9, 7, 6, 7, 5, 3, 5, 4, 2]

# convert to sets
a_set = set(a_list)
b_set = set(b_list)

print(a_set)
print(b_set)

# intersection
intersection = a_set & b_set
print(intersection)

# union
union = a_set | b_set
print(union)

# difference
difference = a_set - b_set
print(difference)

# pair each number with its square 
# list of (1, 2, 3) should become tuple of (1 : 1, 2 : 4, 3 : 9)
"""
c_tuple = tuple(intersection)

for numbers in intersection: 
    sq = numbers**2
    print(f"{numbers} : {sq}")
"""
sq_list = []
for numbers in intersection:
    sq = numbers ** 2
    sq_list.append(sq)

print(sq_list)

res = list(zip(intersection, sq_list))
print(res)
```

4. Create a dictionary representing a student with keys for name, age, major, and GPA. <br>Add a
new key for "graduation_year" and update the GPA. <br>Print all the information in a formatted
way.

```python
student = {
    "name": "Samuel",
    "age": 23,
    "major": "Computer Science",
    "gpa": 2.5,
}

while True:
    x = input("Enter graduation year: ")

    try:
        graduation_year = int(x);
        if graduation_year > 1900:
            break
        else:
            print("Invalid graduation year, please try again.")
    except ValueError:
        print("Wrong input, please try again.")

while True:
    y =  input("Enter GPA: ")
    try: 
        gpa = float(y);
        if gpa >= 0.0 and gpa <= 4.0:
            break
        else: 
            print("Invalid GPA, please try again.")
    except ValueError: 
        print("Wrong input, please try again.")

student["graduation year"] = graduation_year
student["gpa"] = gpa

for key, value in student.items():
    print(f"{key}: {value}")
    
```

5. Use Counter from collections to count the frequency of each letter in a string, ignoring
case. <br>Return the 3 most common letters.

```python
from collections import Counter

#string = input("Enter a string: ")
string = "strings are not lists"
lower = string.lower()

lower.split()
print(lower)
cnt = Counter(lower)


most_common = cnt.most_common(3)
print(string)
print(most_common)

```
