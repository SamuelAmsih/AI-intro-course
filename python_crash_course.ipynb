## 3. Functions

Functions package behavior behind a name. These examples cover required and default parameters, arbitrary positional arguments (`*args`), multiple return values, keyword arguments, and functions passed as values.

Lambdas are small anonymous functions. Prefer a normal `def` when the behavior needs a meaningful name or more than one expression.

```python
def add(a, b):
    return a + b


def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"


result = greet("Alice")


def sum_all(*args):
    return sum(args)


print("add:", add(2, 3))
print("greet:", result)
print(sum_all(1, 2, 3, 4, 5))  # 15
print(sum_all(10, 20))  # 30


def divide_and_remainder(a, b):
    return a // b, a % b


quotient, remainder = divide_and_remainder(17, 5)


def apply_to_one(f):
    return f(1)


y = apply_to_one(lambda x: x + 4)  # 5
print("quotient, remainder:", quotient, remainder)
print("lambda result:", y)


def full_name(first="First-name", last="Last-name"):
    return first + " " + last


print(full_name("Joel", "Scarinius Stävmo"))
print(full_name("Joel"))
print(full_name(last="Scarinius Stävmo"))
```

## 5. Lists

Lists are ordered, mutable collections. Indexing selects one item, slicing creates a new list, and methods such as `append`, `extend`, `insert`, `remove`, and `pop` modify the list.

The final comparison intentionally shows both storing intermediate results and calling functions directly inside an f-string.

```python
numbers = [1, 2, 3, 4, 5]
mixed = [1, "hello", 3.14, True]

first = numbers[0]
print(first)
last = numbers[-1]
print(last)
middle = numbers[1:4]
print(middle)
every_other = numbers[::2]
print(every_other)
print("Mixed:", mixed)

numbers = [1, 2, 3, 4, 5]
print("List:", numbers)
numbers.append(6)
print("List:", numbers)
numbers.extend([7, 8])
print("List:", numbers)
numbers.insert(0, 0)
print("List:", numbers)
numbers.remove(5)
print("List:", numbers)
popped = numbers.pop()
print("List:", numbers)

length = len(numbers)
total = sum(numbers)
maximum = max(numbers)
print(f"popped: {popped}, length: {length}, total: {total}, maximum: {maximum}")

# puts the popped number back
numbers.append(8)

print("Same result without using variables:")
# NOTE: You can also run the functions in the directly in the print statement like this:
print(
    f"popped: {numbers.pop()}, length: {len(numbers)}, total: {sum(numbers)}, maximum: {max(numbers)}"
)
```

## 6. Tuples and unpacking

Tuples are ordered and immutable. They are useful for fixed records, returning multiple values, unpacking assignments, and dictionary keys when all contained values are hashable.

The comma creates a tuple, which is why a one-item tuple must be written as `(42,)`.

```python
point = (3, 4)
person = ("Alice", 25, "Engineer")
single_item = (42,)

x, y = point
print(f"The points values are: X = {x}, Y = {y}")
name, age, job = person
print(f"The persons job is: {job}")

a, b = 1, 2
print(f"Before swapping: a = {a} and b = {b}")
a, b = b, a
print(f"After swapping: a = {a} and b = {b}")
print(f"Point: {point}, Person: {person}, Single item: {single_item}")


def get_name_age():
    return "Bob", 30


name, age = get_name_age()
locations = {
    (0, 0): "origin",
    (1, 0): "right",
    (0, 1): "up",
}
print(f"Name: {name} & Age: {age}")
print(f"Locations: {locations}")
```

## 7. Dictionaries, defaultdict, and Counter

Dictionaries map unique hashable keys to values. `defaultdict` creates missing values through a factory, while `Counter` is specialized for counting hashable items.

Use `mapping[key]` when a missing key is an error. Use `mapping.get(key, default)` when absence is expected.

```python
student = {
    "name": "Alice",
    "age": 20,
    "grades": [85, 92, 78],
}
print("Initial student:", student)

student = dict(name="Alice", age=20)
name = student["name"]
age = student.get("age", 0)
student["major"] = "CS"
print(name, age, student)

student = dict(name="Alice", age=20)
keys = student.keys()
values = student.values()
items = student.items()
print("keys:", list(keys))
print("values:", list(values))
print("items:", list(items))

for key, value in student.items():
    print(f"{key}: {value}")

squares = {x: x**2 for x in range(5)}
print("squares:", squares)

regular_dict = {}
print("regular_dict:", regular_dict)
# regular_dict["missing"] would raise KeyError.

dd_int = defaultdict(int)
print("default int:", dd_int["missing"])

dd_list = defaultdict(list)
dd_list["fruits"].append("apple")
dd_list["fruits"].append("banana")
print("default list:", dd_list)

words = ["apple", "banana", "apricot"]
by_first_letter = defaultdict(list)
for word in words:
    by_first_letter[word[0]].append(word)
print("grouped:", dict(by_first_letter))

text = "hello world"
char_count = Counter(text)
print("char count:", char_count)

words = ["apple", "banana", "apple", "cherry"]
word_count = Counter(words)
print("most common:", word_count.most_common(2))

c1 = Counter(["a", "b", "c", "a"])
c2 = Counter(["a", "b", "b"])
print("counter addition:", c1 + c2)
print("counter subtraction:", c1 - c2)

word_count.update(["apple", "date"])
print("updated apple count:", word_count["apple"])
```

## 8. Sets

Sets store unique hashable values and support mathematical operations such as union, intersection, and difference. Set iteration order should not be used as a sequencing guarantee.

`remove` raises `KeyError` when an item is absent; `discard` does not.

```python
colors = {"red", "green", "blue"}
numbers = set([1, 2, 3, 3, 2, 1])
empty_set = set()

if "red" in colors:
    print("Red is a color")

colors.add("yellow")
# KeyError if not present
colors.remove("red")
# No error if not present
colors.discard("purple")
print(f"Colors: {colors}, Numbers: {numbers}, Empty Set: {empty_set}")

set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}
union = set1 | set2
intersection = set1 & set2
difference_1 = set1 - set2
difference_2 = set2 - set1
symmetric_diff = set1 ^ set2
# NOTE: this is a set and not a list and that is why output is unordered. You can read more about how sets work out online.
even_squares = {x**2 for x in range(10) if x % 2 == 0}
print("union:", union)
print("intersection:", intersection)
print("difference 1:", difference_1)
print("difference 2:", difference_2)
print("symmetric difference:", symmetric_diff)
print("even squares:", even_squares)
```

## 9. Conditionals and truth values

Conditionals choose which code runs. Python treats empty containers, zero, `None`, and `False` as falsy; many other values are truthy.

The examples compare a compound condition with the more idiomatic chained comparison `18 <= age <= 69`.

```python
age = 18
if age >= 18:
    print("You can vote")
elif age >= 16:
    print("You can drive")
else:
    print("You're still young")

values = [0, "", [], {}, None, False, "hello", [1, 2]]
for val in values:
    if val:
        # NOTE: !r creates '' encapsulating string outputs.
        print(f"{val!r} is truthy")
    else:
        print(f"{val!r} is falsy")

if age >= 18 and age <= 69:
    print("Working age")

a, b = 2, 3
status = "adult" if age >= 18 else "minor"
print("status", status)
max_val = a if a > b else b
print("max:", max_val)
if 18 <= age <= 69:
    print("Working age")
```

## 10. Loops and loop control

Use `for` to iterate over items and `while` to repeat while a condition remains true. `enumerate` adds indexes, `zip` pairs iterables, `continue` skips to the next iteration, and `break` exits the nearest loop.

A loop's `else` block runs only when the loop finishes without `break`.

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
print("0 - 4")
for i in range(5):
    print(i)
print("Every other 1-10")
for i in range(1, 10, 2):
    print(i)
print('Using "enumerate":')
for i, fruit in enumerate(fruits):
    print(f"{i}: {fruit}")

names = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 35]
print('Using "zip":')
for name, age in zip(names, ages):
    print(f"{name} is {age} years old")

print("Counting with a while loop:")
count = 0
while count < 5:
    print(count)
    count += 1

print('For loop with "break" and "continue"')
for i in range(10):
    if i == 3:
        continue  # skips rest of that iteration of the loop
    if i == 7:
        break
    print(i)

numbers = [1, 3, 5, 7, 9]
target = 4
print('For loop with "else"')
for num in numbers:
    if num == target:
        print("Found target!")
        break
else:
    print("Target not found")

for i in range(4):
    if i == 2:
        pass  # just a placeholder that does nothing, loop will just continue as normal.
    else:
        print(i)
```

## 12. Comprehensions and generators

Comprehensions concisely build lists, dictionaries, and sets. A generator expression produces items lazily and is consumed as it is iterated.

Readable loops are often better than deeply nested comprehensions. Both approaches are shown so learners can compare them.

```python
squares = [x**2 for x in range(10)]
even_squares = [x**2 for x in range(10) if x % 2 == 0]
words = ["hello", "world", "python"]
lengths = [len(word) for word in words]
upper_words = [word.upper() for word in words]
print(squares, even_squares, lengths, upper_words, sep="\n")

matrix = [[i + j for j in range(3)] for i in range(3)]
nested = [[1, 2], [3, 4], [5, 6]]
flat = [item for sublist in nested for item in sublist]
print("matrix:", matrix)
print("flat:", flat)

print("Above and below is just two different ways of doing the same thing")

flat = []
for sublist in nested:
    for item in sublist:
        flat.append(item)
print("flat using loops:", flat)

word_lengths = {word: len(word) for word in ["hello", "world", "python"]}
squares_dict = {x: x**2 for x in range(5)}
unique_lengths = {len(word) for word in ["hello", "hi", "world", "hey"]}
print(word_lengths, squares_dict, unique_lengths, sep="\n")

sum_squares = sum(x**2 for x in range(1_000_000))
print("sum squares:", sum_squares)
data = ["1", "2", "3", "4", "5"]
numbers_list = [int(x) for x in data]
print(f"List: {numbers_list}")
numbers_set = {int(x) for x in data}
print(f"Set: {numbers_set}")
numbers_dict = {x: int(x) for x in data}
print(f"Dict: {numbers_dict}")
numbers_gen = (int(x) for x in data)
print(f"Generator: {numbers_gen}")
```

## 13. Classes and object-oriented programming

A class defines a new type. Instance attributes belong to each object; class attributes are shared through the class. Methods receive the instance as `self` by convention.

Python does not enforce `public`, `protected`, and `private` in the same way as languages such as C#. A leading underscore communicates non-public intent. A double leading underscore triggers name mangling, but it is not an access-control boundary.

Type hints below improve readability and tooling while leaving runtime behavior unchanged.

```python
class Student:
    school = "University of Python"

    def __init__(self, name: str, age: int, major: str) -> None:
        self.name = name
        self.age = age
        self.major = major
        self.grades: list[float] = []

    def add_grade(self, grade: float) -> None:
        self.grades.append(grade)

    def get_gpa(self) -> float:
        if not self.grades:
            return 0.0
        return sum(self.grades) / len(self.grades)

    def __str__(self) -> str:
        return f"{self.name}, {self.age}, {self.major}"


alice = Student("Alice", 20, "Computer Science")
bob = Student("Bob", 22, "Mathematics")
alice.add_grade(85)
alice.add_grade(92)
print(alice)
print(bob)
print(f"Alice's GPA: {alice.get_gpa()}")
```

## 14. Sorting

`sorted(iterable)` returns a new list. `list.sort()` changes an existing list in place and returns `None`. The `key` function computes the value used for comparison; it is usually clearer than writing custom comparison logic.

```python
class GradedStudent:
    """Renamed from the second slide's Student class to avoid overwriting Student."""

    def __init__(self, name, grade):
        self.name = name
        self.grade = grade


numbers = [3, 1, 4, 1, 5, 9]
print(sorted(numbers))
print(sorted(numbers, reverse=True))

more_numbers = [2, 3, 5, 6, 8, 11, 3, 1, 4, 1, 5, 9]

more_numbers.sort()

print("Sorted using in-place function .sort(): ", more_numbers)

words = ["python", "java", "c", "javascript"]
print(sorted(words, key=len))

pairs = [("a", 2), ("b", 1), ("c", 3)]
print(sorted(pairs, key=lambda x: x[1]))

students = [GradedStudent("Alice", 85), GradedStudent("Bob", 90)]
top_students = sorted(students, key=lambda student: student.grade, reverse=True)
print([(student.name, student.grade) for student in top_students])
```

## 15. Generators in more depth

A function containing `yield` returns a generator. Execution pauses after each yielded value and resumes when the next value is requested. This is useful for streams or large sequences that should not be stored entirely in memory.

An infinite generator, such as `fibonacci`, must be paired with a stopping mechanism. Here, `itertools.islice` safely requests only ten values.

```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1


print("countdown:", list(countdown(5)))

squares_gen = (x**2 for x in range(1_000_000))
first_square = next(squares_gen)
second_square = next(squares_gen)
print("first squares:", first_square, second_square)


def fibonacci():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b


import itertools

fib = fibonacci()
first_10_fibs = list(itertools.islice(fib, 10))
print("first 10 Fibonacci numbers:", first_10_fibs)

list_comp = [x**2 for x in range(1000)]
gen_exp = (x**2 for x in range(1000))
print("list length:", len(list_comp), "first generated value:", next(gen_exp))
```

## 16. Inheritance and Python's data model

Inheritance lets one class extend another. `super()` delegates initialization to the parent class. Special methods such as `__len__` and `__getitem__` integrate user-defined objects with built-in operations like `len(obj)` and indexing.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        return f"Hi, I'm {self.name}"


class InheritedStudent(Person):
    """Student inheritance example from a commented-out slide."""

    def __init__(self, name, age, student_id):
        super().__init__(name, age)
        self.student_id = student_id
        self.courses = []

    def introduce(self):
        return f"Hi, I'm {self.name}, student ID: {self.student_id}"

    def __len__(self):
        return len(self.courses)

    def __getitem__(self, index):
        return self.courses[index]


student = InheritedStudent("Alice", 20, "S12345")
print(student.introduce())
student.courses = ["Math", "Physics", "CS"]
print(len(student))
print(student[0])
```

## 17. Important modern additions for a well-rounded crash course

The original material covers the core language well. The following compact examples add topics that are now common in practical Python:

- **type hints** for readers and static-analysis tools;
- **dataclasses** for data-focused classes;
- **`pathlib`** for object-oriented filesystem paths;
- **structural pattern matching** for shape-based branching;
- **context managers** for reliable resource cleanup.

These features supplement rather than replace the earlier examples.

```python
from dataclasses import dataclass, field
from pathlib import Path
from typing import Iterable


def average(values: Iterable[float]) -> float:
    """Return the arithmetic mean of a non-empty iterable."""
    collected = list(values)
    if not collected:
        raise ValueError("average() needs at least one value")
    return sum(collected) / len(collected)


@dataclass(slots=True)
class Course:
    """A small data-focused class with a safe per-instance list."""

    code: str
    title: str
    students: list[str] = field(default_factory=list)


course = Course("PY101", "Python Crash Course")
course.students.append("Alice")
print(course)
print("Average:", average([85, 92, 78]))

# pathlib composes paths without manually inserting separators.
notebook_path = Path.cwd() / "python_crash_course_notebook.ipynb"
print("Example path:", notebook_path)


def describe(value: object) -> str:
    """Classify a few common shapes using structural pattern matching."""
    match value:
        case {"name": str(name), "age": int(age)}:
            return f"person record for {name}, age {age}"
        case [first, *rest]:
            return f"sequence beginning with {first!r}; {len(rest)} item(s) remain"
        case None:
            return "no value"
        case _:
            return "something else"


print(describe({"name": "Alice", "age": 25}))
print(describe([10, 20, 30]))
```
