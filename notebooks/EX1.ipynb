7. Write a program that takes a test score (0-100) and assigns a letter grade based on the
following scale: <br>A (90-100), B (80-89), C (70-79), D (60-69), F (below 60)

```python
while True: 
    try:
        score = input("Test grader. \n To quit program enteter 'EXIT' \n Enter test score: ")

        if score == "Exit":
            print("Exiting...")
            break

        score = int(score)

        invalid_score = score > 100 or score < 0
        
        if invalid_score == True:     
            print("Invalid score, please try again.")
            continue

        
        if score > 89:
            grade = "A"
        elif score > 79:
            grade = "B"
        elif score > 69:
            grade = "C"
        elif score > 59:
            grade = "D"
        else: 
            grade = "F"
        
        print(f"Your grade is: {grade}")
        break
             
    except ValueError: 
        print("Invalid input, please try again.")
```

10. Create a list of numbers from 1 to 20. <br>Use a loop to find all even numbers. <br>Then use a list
comprehension to create a list of squares of odd numbers from the original list.

```python
y = list(range(1,21))

print(y)

even = []
odd  = []

for nums in y:
    if nums % 2 == 0:
        even.append(nums)
    else: 
        odd.append(nums)

# comprehension
# output            collection            condition
# x**3          for this collection     in this situation
odd_square = [x**2 for x in y if x%2 == 1]


print(even)
print(odd)
print(odd_square)


```

11. Create a function that counts the number of vowels in a string using a loop and a set. Handle
edge cases like empty strings.

```python
def vowel_count(string):
    vowels = {'A', 'a', 'E', 'e', 'I', 'i', 'O', 'o', 'U', 'u' }

    y = string.split()

    num_vowels = 0

    for word in y:
        #print(word)
        for letter in word:
            #print(letter)
                if letter in vowels:
                    num_vowels += 1

    return num_vowels

string = "Roll back to kitchen"

vowel_count(string)

    
```

12. Write a function called calculate_statistics that takes a list of numbers and returns
a dictionary containing the minimum, maximum, sum, and average of the numbers. Use
built-in functions for statistics. Print the resulting dictionary.

```python
def calculate_statistics(x):
    d = {
        "min": min(x), 
        "max": max(x),
        "sum": sum(x),
        "avg": sum(x) / len(x)
    }

    return d


x = [1, 2, 3, 4]
for k, v in calculate_statistics(x).items():
    print(f"{k} : {v}")
```
